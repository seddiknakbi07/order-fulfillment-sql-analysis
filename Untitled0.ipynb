```python
import sqlite3
import pandas as pd

conn = sqlite3.connect("fulfillment.db")

for name in ["Products", "Customers", "Orders", "Order_Lines", "Shipments"]:
    df = pd.read_csv(f"{name}.csv")
    df.to_sql(name, conn, if_exists="replace", index=False)

print("Tables loaded:", pd.read_sql("SELECT name FROM sqlite_master WHERE type='table'", conn))
```

```python
pd.read_sql("""
SELECT *
FROM Orders
LIMIT 5;
""", conn)
```

```python
pd.read_sql("""
SELECT
    ol.Order_Line_ID,
    ol.Order_ID,
    o.Customer_ID,
    o.Order_Date,
    o.Promised_Date,
    ol.SKU_ID,
    ol.Qty_Ordered,
    ol.Qty_Shipped,
    s.Carrier,
    s.Delivery_Date
FROM Order_Lines ol
JOIN Orders o ON ol.Order_ID = o.Order_ID
JOIN Shipments s ON ol.Order_ID = s.Order_ID
LIMIT 10;
""", conn)
```

```python
pd.read_sql("""
SELECT
    ROUND(100.0 * AVG(CASE WHEN ol.Qty_Shipped >= ol.Qty_Ordered THEN 1 ELSE 0 END), 1) AS Fill_Rate_Pct,
    ROUND(100.0 * AVG(CASE WHEN s.Delivery_Date <= o.Promised_Date THEN 1 ELSE 0 END), 1) AS OnTime_Rate_Pct,
    ROUND(100.0 * AVG(CASE WHEN ol.Qty_Shipped >= ol.Qty_Ordered
                            AND s.Delivery_Date <= o.Promised_Date THEN 1 ELSE 0 END), 1) AS OTIF_Pct
FROM Order_Lines ol
JOIN Orders o ON ol.Order_ID = o.Order_ID
JOIN Shipments s ON ol.Order_ID = s.Order_ID;
""", conn)
```

```python
pd.read_sql("""
SELECT
    s.Carrier,
    COUNT(*) AS Num_Lines,
    ROUND(100.0 * AVG(CASE WHEN s.Delivery_Date <= o.Promised_Date THEN 1 ELSE 0 END), 1) AS OnTime_Rate_Pct,
    ROUND(100.0 * AVG(CASE WHEN ol.Qty_Shipped >= ol.Qty_Ordered
                            AND s.Delivery_Date <= o.Promised_Date THEN 1 ELSE 0 END), 1) AS OTIF_Pct
FROM Order_Lines ol
JOIN Orders o ON ol.Order_ID = o.Order_ID
JOIN Shipments s ON ol.Order_ID = s.Order_ID
GROUP BY s.Carrier
ORDER BY OTIF_Pct ASC;
""", conn)
```

```python
pd.read_sql("""
SELECT
    strftime('%Y-%m', o.Order_Date) AS Order_Month,
    COUNT(*) AS Num_Lines,
    ROUND(100.0 * AVG(CASE WHEN ol.Qty_Shipped >= ol.Qty_Ordered
                            AND s.Delivery_Date <= o.Promised_Date THEN 1 ELSE 0 END), 1) AS OTIF_Pct
FROM Order_Lines ol
JOIN Orders o ON ol.Order_ID = o.Order_ID
JOIN Shipments s ON ol.Order_ID = s.Order_ID
GROUP BY Order_Month
ORDER BY Order_Month;
""", conn)
```

```python
import matplotlib.pyplot as plt

carrier_df = pd.read_sql("""
SELECT
    s.Carrier,
    ROUND(100.0 * AVG(CASE WHEN ol.Qty_Shipped >= ol.Qty_Ordered
                            AND s.Delivery_Date <= o.Promised_Date THEN 1 ELSE 0 END), 1) AS OTIF_Pct
FROM Order_Lines ol
JOIN Orders o ON ol.Order_ID = o.Order_ID
JOIN Shipments s ON ol.Order_ID = s.Order_ID
GROUP BY s.Carrier
ORDER BY OTIF_Pct ASC;
""", conn)

colors = ["#DD8452" if c == "Canada Post" else "#4C72B0" for c in carrier_df["Carrier"]]

plt.figure(figsize=(7,5))
plt.bar(carrier_df["Carrier"], carrier_df["OTIF_Pct"], color=colors)
plt.ylabel("OTIF %")
plt.title("On-Time-In-Full Rate by Carrier")
plt.ylim(0, 100)
for i, v in enumerate(carrier_df["OTIF_Pct"]):
    plt.text(i, v + 2, f"{v}%", ha="center")
plt.tight_layout()
plt.savefig("otif_by_carrier.png", dpi=150)
plt.show()
```

```python
month_df = pd.read_sql("""
SELECT
    strftime('%Y-%m', o.Order_Date) AS Order_Month,
    ROUND(100.0 * AVG(CASE WHEN ol.Qty_Shipped >= ol.Qty_Ordered
                            AND s.Delivery_Date <= o.Promised_Date THEN 1 ELSE 0 END), 1) AS OTIF_Pct
FROM Order_Lines ol
JOIN Orders o ON ol.Order_ID = o.Order_ID
JOIN Shipments s ON ol.Order_ID = s.Order_ID
GROUP BY Order_Month
ORDER BY Order_Month;
""", conn)

plt.figure(figsize=(7,5))
plt.plot(month_df["Order_Month"], month_df["OTIF_Pct"], marker="o", color="#4C72B0")
plt.ylabel("OTIF %")
plt.title("OTIF Rate by Month")
plt.ylim(0, 100)
plt.tight_layout()
plt.savefig("otif_by_month.png", dpi=150)
plt.show()
```
